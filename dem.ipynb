```python
import pygetwindow as gw


def get_system_state():

    try:

        win = gw.getActiveWindow()

        if not win:
            return {
                "active_app":None,
                "window_title":None
            }

        title=win.title

        return {
            "active_app": title.split("-")[-1].strip()
                if "-" in title
                else title,

            "window_title": title
        }

    except Exception:

        return {
            "active_app":None,
            "window_title":None
        }
```

```python
import psutil
import win32gui
import win32process

def get_running_apps():
    # 1. Get all window handles that are currently visible
    def callback(hwnd, hwnds):
        if win32gui.IsWindowVisible(hwnd) and win32gui.GetWindowText(hwnd):
            hwnds.append(hwnd)
    
    visible_hwnds = []
    win32gui.EnumWindows(callback, visible_hwnds)
    
    # 2. Get the unique Process IDs (PIDs) linked to those windows
    app_pids = set()
    for hwnd in visible_hwnds:
        _, pid = win32process.GetWindowThreadProcessId(hwnd)
        app_pids.add(pid)
    
    # 3. Use psutil to get the names for those PIDs
    for pid in app_pids:
        try:
            print(psutil.Process(pid).name())
        except (psutil.NoSuchProcess, psutil.AccessDenied):
            continue

get_running_apps()
```

```python
import win32gui
import win32process
import psutil


APP_PROCESS = {
   "chrome":"chrome.exe",
   "vscode":"Code.exe",
   "notepad":"notepad.exe"
}


def focus_app(app):

    target=APP_PROCESS.get(
       app.lower()
    )

    if not target:
        raise Exception(
          "unknown app"
        )


    target_pids=[]


    for proc in psutil.process_iter(
      ['pid','name']
    ):
        try:
            if (
              proc.info["name"]
              == target
            ):
                target_pids.append(
                    proc.info["pid"]
                )
        except:
            pass



    if not target_pids:
        raise Exception(
           f"{app} not running"
        )


    def enum_handler(hwnd,_):

        if not win32gui.IsWindowVisible(
            hwnd
        ):
            return

        _,pid=win32process.GetWindowThreadProcessId(
            hwnd
        )

        if pid in target_pids:

            win32gui.ShowWindow(
                hwnd,
                5
            )

            win32gui.SetForegroundWindow(
                hwnd
            )

            raise StopIteration


    try:
        win32gui.EnumWindows(
            enum_handler,
            None
        )

    except StopIteration:
        return {
          "status":"success",
          "action":"focus_app"
        }


    raise Exception(
       "No window found"
    )
    
print(focus_app("chrome"))
```
